```python
#### used to for the vector embeddings
from sentence_transformers import SentenceTransformer

model_path = r"D:\ollamaModels\all-mpnet-base-v2"

encoder = SentenceTransformer(model_path)

vectors = encoder.encode(df["text"].tolist())

vectors.shape
```

```python
dim = vectors.shape[1]
dim
```

```python
import faiss

index = faiss.IndexFlatL2(dim)
index
```

```python
index.add(vectors)
```

```python
# search_query = "I want to buy a polo t-shirt"

search_query = "An appla a day keeps doctor away"

vec = encoder.encode(search_query)
vec.shape
```

```python
# [vec]
```

```python
import numpy as np
svec = np.array(vec).reshape(1,-1)
svec.shape
```

```python
### k stands for the number of nearest neighbors you want to retrieve.
distances, I = index.search(svec, k=2)
```
